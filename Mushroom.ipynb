```python
mushroom.drop(columns=['veil-type'],axis=  1,inplace=True)
```

```python
def Label_enc(column):
  LE=LabelEncoder()
  LE.fit(column)
  print(column.name,LE.classes_)
  return LE.transform(column)
```

```python
for i in mushroom.columns:
  mushroom[str(i)] = Label_enc(mushroom[str(i)])
```

```python
mushroom.head()
```

```python
x = mushroom.drop(columns=['class'], axis=1)
y=mushroom['class']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size = 0.2)
```

```python
model=RandomForestClassifier()
model.fit(x_train, y_train)
y_predict = model.predict(x_test)
```

```python
print("The accuracy of the model is", accuracy_score(y_predict, y_test))
```
